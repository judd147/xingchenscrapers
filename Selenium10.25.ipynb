```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.edge.service import Service
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.wait import WebDriverWait
from selenium.webdriver import ActionChains
from selenium.webdriver.common.actions.wheel_input import ScrollOrigin
from selenium.webdriver.common.keys import Keys

def strip_parent(string):
    return string.split('(')[1].split(')')[0]
```

```python
#driver_path = r'D:\edgedriver_win64\msedgedriver.exe'
driver_path = r'D:\EdgeDriver\msedgedriver.exe'
driver = webdriver.Edge(service=Service(executable_path=driver_path))
driver.implicitly_wait(10)
driver.get("https://www.sofascore.com/")
```

```python
#显示赔率
showodds = driver.find_element(By.CLASS_NAME,'slider')
showodds.click()
```

```python
#获取当前日期
current_date = driver.find_elements(By.CLASS_NAME,'sc-egiyK.lbdavM')
for item in current_date:
    print(item.text)
```

```python
#切换日期
dates = driver.find_elements(By.CLASS_NAME,'sc-egiyK.dbNpeU')
for date in dates:
    if date.text == '22':
        print('yes')
        date.click()
```

```python
#打开新标签页获取盘口
element = driver.find_element(By.CLASS_NAME,'sc-bdvvtL.izeUUn')
element.send_keys(Keys.CONTROL + Keys.RETURN)
driver.switch_to.window(driver.window_handles[1])
#存储当前比赛数据
handicap = driver.find_elements(By.CLASS_NAME,'sc-egiyK.bPBNjz')
print('比赛：', handicap[0].text.split(')')[1]+'-'+handicap[1].text.split(')')[1])
print('盘口：', strip_parent(handicap[0].text.split(' ')[0]))
value = driver.find_elements(By.CLASS_NAME,'sc-egiyK.iJPpi')
print(value[0].text,':',value[2].text)
print(value[1].text,':',value[3].text)
#回到主页面
driver.close()
driver.switch_to.window(driver.window_handles[0])
```
