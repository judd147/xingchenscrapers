```python
def init_service(headless):
    driver_path = "/Users/zhangliyao/Downloads/chromedriver"
    # Mobile Device
    mobile_emulation = {"deviceName": "iPhone 12 Pro"}

    # Initialize Driver
    chrome_options = webdriver.ChromeOptions()
    chrome_options.add_argument('--enable-automation')
    if headless:
        chrome_options.add_argument('--headless')
    chrome_options.add_experimental_option("mobileEmulation", mobile_emulation)
    try:
        driver = webdriver.Chrome(service=Service(executable_path=ChromeDriverManager().install()), options=chrome_options)
    except:
        driver = webdriver.Chrome(service=Service(executable_path=driver_path), options=chrome_options)
        

    # Open web page
    driver.get('https://xczy.sp1x2.net/xczy-web/')
    driver.fullscreen_window()
    return driver

def login(driver):
    try:
        driver.find_element(By.CLASS_NAME, 'van-dialog__cancel').click()
    except:
        pass
    # Input the phone number
    phone_input = WebDriverWait(driver, 10).until(EC.presence_of_element_located(
        (By.CLASS_NAME, 'van-field__control'))).send_keys('18566258659')

    # Input the password
    password_input = driver.find_elements(By.CLASS_NAME, 'van-field__control')[1]
    password_input.send_keys('990311')

    # Login button
    login_button = WebDriverWait(driver, 10).until(
        EC.element_to_be_clickable((By.CLASS_NAME, 'login-btn')))
    login_button.click()
    
def pct_to_float(pct):
    '''
    Convert percent String to float
    '''
    if isinstance(pct, str):
        return float(pct.strip('%'))/100
    return pct

def get_matches(driver, date, data):
    loading = True
    while loading:
        try:
            spinner = driver.find_element(By.CLASS_NAME, 'van-loading__text')
            time.sleep(3)
        except:
            loading = False

    try:
        matches = driver.find_element(By.CLASS_NAME, 'list-ul').find_elements(By.CSS_SELECTOR, 'li')
    except:
        return
    
    for match in matches:
        driver.execute_script("arguments[0].scrollIntoView({ block: 'center' });", match) # scroll to the target
        # 基本信息
        kickoff_time = match.find_element(By.CLASS_NAME, 'time').text
        datetime_str = date + ' ' + kickoff_time
        datetime_obj = datetime.strptime(datetime_str, "%Y年%m月%d %H:%M").strftime("%Y-%m-%d %H:%M") # 开球时间

        league = match.find_element(By.CLASS_NAME, 'league').text # 联赛

        teams = match.find_elements(By.CLASS_NAME, 't-name')
        score_FT = match.find_element(By.CLASS_NAME, 'score').text
        match_text = teams[0].text + score_FT + teams[1].text # 比赛

        try:
            score_HT = match.find_element(By.CLASS_NAME, 'left.right').text.split(" ")[1] # 半场比分
        except:
            continue

        goals_array = score_FT.split('-')
        goals_FT = int(goals_array[0]) + int(goals_array[1]) # 全场进球数

        goals_array = score_HT.split(':')
        goals_HT = int(goals_array[0]) + int(goals_array[1]) # 半场进球数

        handicap_goals = match.find_elements(By.CLASS_NAME, 'asian')[1].text # 大小盘

        #print(datetime_obj, league, match_text, score_HT, goals_FT, goals_HT, handicap_goals)

        # 火线数据
        expand = match.find_element(By.CLASS_NAME, 'fire')
        if expand.text == '展开':
            expand.click()
        else:
            continue
        time.sleep(1)
        expand = match.find_element(By.CLASS_NAME, 'fire')
        expand.click()

        contents = match.find_elements(By.CLASS_NAME, 'fire-content-main')

        HT_O_05 = np.nan # 半场大于0.5球
        HT_U_05 = np.nan # 半场小于0.5球
        FT_O_25 = np.nan # 全场大于2.5球
        FT_U_25 = np.nan # 全场小于2.5球
        FG_Home = np.nan # 主队先进球
        FG_Away = np.nan # 客队先进球
        G_A_80m = np.nan # 80分钟后进球
        A_winner = '' # 亚盘获胜者
        AW_prob = np.nan # 亚盘赢指
        FT_GG_B = np.nan # 双方进球
        FT_O_05 = np.nan # 破蛋概率
        num_G_O = np.nan # 全场进球单数
        num_G_E = np.nan # 全场进球双数
        FT_O_15 = np.nan # 全场大于1.5球
        SH_O_15 = np.nan # 下半场大于1.5球
        HT_GG_B = np.nan # 上半场双方进球
        G_B_15m = np.nan # 前15分钟进球

        MS_HT_05 = np.nan # 半场大小遗漏
        MS_FT_25 = np.nan # 全场大小遗漏
        MS_Asian = np.nan # 亚盘遗漏
        MS_FT_GG = np.nan # 双方进球遗漏
        MS_NG_OE = np.nan # 单双数遗漏
        MS_FT_15 = np.nan # 全场大1.5遗漏

        for content in contents:
            probability = pct_to_float(content.find_element(By.CLASS_NAME, 'f-c-p').text) # 概率

            detail = content.find_element(By.CLASS_NAME, 'f-c-m')
            event = detail.text # 事件

            history = detail.find_elements(By.CSS_SELECTOR, 'span')
            last_seven = content.find_element(By.CLASS_NAME, 'hit') 
            misses = last_seven.find_elements(By.CLASS_NAME, 'hit-f') 
            num_misses = len(misses) if len(history) > 0 else np.nan # 近七场未命中次数
            miss_streak = 0 # 遗漏数
            for hit in history:
                if hit.get_attribute('class') == 'hit-f':
                    miss_streak += 1
                else:
                    miss_streak = 0
            miss_streak = miss_streak if len(history) > 0 else np.nan

            if event.__contains__('亚指'):
                # 亚洲盘
                handicap = event.split(' ')[0]
                if not handicap.startswith('-') and handicap != '0':
                    handicap = '+'+handicap

            if event == '半场大于0.5球概率':
                HT_O_05 = probability
                MS_HT_05 = miss_streak
            elif event == '半场小于0.5球概率':
                HT_U_05 = probability
                MS_HT_05 = miss_streak
            elif event == '全场大于2.5球概率':
                FT_O_25 = probability
                MS_FT_25 = miss_streak
            elif event == '全场小于2.5球概率':
                FT_U_25 = probability
                MS_FT_25 = miss_streak
            elif event == '主队先进球概率':
                FG_Home = probability
            elif event == '客队先进球概率':
                FG_Away = probability
            elif event == '80分钟后有进球概率':
                G_A_80m = probability
            elif '赢指数' in event:
                if '主队' in event:
                    A_winner = '主'
                elif '客队' in event:
                    A_winner = '客'
                AW_prob = probability
                MS_Asian = miss_streak
            elif event == '两队都进球概率':
                FT_GG_B = probability
                MS_FT_GG = miss_streak
            elif event == '破蛋概率':
                FT_O_05 = probability
            elif event == '全场进球单数':
                num_G_O = probability
                MS_NG_OE = miss_streak
            elif event == '全场进球双数':
                num_G_E = probability
                MS_NG_OE = miss_streak
            elif event == '全场大于1.5球':
                FT_O_15 = probability
                MS_FT_15 = miss_streak
            elif event == '下半场大于1.5球':
                SH_O_15 = probability
            elif event == '上半场两队都有进球':
                HT_GG_B = probability
            elif event == '前15分钟有进球概率':
                G_B_15m = probability

            #print(probability, event, num_misses, miss_streak)
            
        # 进球时间
        try:
            match.find_element(By.CLASS_NAME, 'event-flod').click()
            timelines = match.find_elements(By.CLASS_NAME, 'events-timeline')
            goals = []
            for timeline in timelines:
                try:
                    timeline.find_element(By.CLASS_NAME, 'ball')
                    minute = timeline.find_element(By.CLASS_NAME, 'event-center').text.split("'")[0]
                    goals.append(minute)
                except:
                    continue
            goals_str = ', '.join(map(str, goals))
            #print(goals_str)
        except:
            goals_str = ''

        data.append([datetime_obj, league, match_text, score_HT, goals_FT, goals_HT, handicap, handicap_goals, 
                     HT_O_05, HT_U_05, FT_O_25, FT_U_25, FG_Home, FG_Away, G_A_80m, A_winner, AW_prob, FT_GG_B,
                     FT_O_05, num_G_O, num_G_E, FT_O_15, SH_O_15, HT_GG_B, G_B_15m, MS_HT_05, MS_FT_25, MS_Asian,
                     MS_FT_GG, MS_NG_OE, MS_FT_15, goals_str])
        #break
        
def fetch_detail(driver, league, data):
    hasPrev = True
    while hasPrev:
        try:
            driver.find_element(By.CLASS_NAME, "Button.RVwfR").click()
        except:
            pass

        container = driver.find_element(By.CLASS_NAME, 'Box.hIovvg')
        matches = container.find_elements(By.CLASS_NAME, "js-list-cell-target")
        for match in matches:
            info_list = match.text.split('\n')
            date = info_list[0]
            code = info_list[1]
            home = info_list[2]
            away = info_list[3]
            if code != 'FT' and code != 'AET' and code != 'AP':
                #print('error match:', home+'-'+away, code)
                continue
            H = info_list[4]
            try:
                A = info_list[6]
            except:
                A = info_list[5]
            
            match.click()
            time.sleep(2)

            try:
                driver.find_element(By.CLASS_NAME, 'dZNeJi.bnpRyo')
            except:
                match.click()
                time.sleep(1)

            try:
                goal_detail = driver.find_element(By.CLASS_NAME, 'bPpPiN.TrPCG').text
            except:
                goal_detail = ''

            goals = [int(goal) for goal in re.findall(r'\d+(?=\'\s*,|\')', goal_detail)]
            goals.sort()
            goals = [x for x in goals if x <= 90]
            goals_str = ', '.join(map(str, goals))

            try:
                date_str = datetime.strptime(date, '%d/%m/%y').strftime('%Y-%m-%d')
            except:
                continue
            #print(date_str, code, home, H+'-'+A, away)

            data.append([date_str, league, home, H+'-'+A, away, goals_str])
            time.sleep(1)

        previous = driver.find_element(By.CLASS_NAME, 'Button.iCnTrv')
        if previous.text == 'PREVIOUS':
            hasPrev = True
            previous.click()
        else:
            hasPrev = False
        time.sleep(1)
        print('已获取', len(data), '场比赛')
        
def count_goals(score):
    goals = score.split('-')
    return int(goals[0]) + int(goals[1])

def check_errors(df):
    #df['进球时间'] = df['进球时间'].str.replace('nan', '')
    # 检查重复值 排序
    df['主队'] = df['主队'].str.replace('x2', '')
    df['客队'] = df['客队'].str.replace('x2', '')
    df['主队'] = df['主队'].str.replace('x3', '')
    df['客队'] = df['客队'].str.replace('x3', '')

    df.drop_duplicates(subset=['开球时间', '联赛', '主队', '比分', '客队'], keep='last', inplace=True)
    print(len(df))

    df.sort_values(by=['开球时间'], inplace=True, kind='stable', ignore_index=True)

    # 标记不完整进球时间比赛
    df['num_goals'] = df['比分'].apply(count_goals)
    df['goals_recorded'] = df['进球时间'].str.split(',').apply(lambda x: len(x) if x[0] != '' else 0)
    df['error'] = df['num_goals'] != df['goals_recorded']
```
