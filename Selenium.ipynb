```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.edge.service import Service
```

```python
#https://developer.microsoft.com/en-us/microsoft-edge/tools/webdriver/
driver_path = r'D:\edgedriver_win64\msedgedriver.exe'
driver = webdriver.Edge(service=Service(executable_path=driver_path))
driver.get("https://www.sofascore.com/")
```

```python
showodds = driver.find_element(By.CLASS_NAME,'slider')
showodds.click()
```

```python
matches = driver.find_elements(By.CLASS_NAME,'ReactVirtualized__Grid__innerScrollContainer')
for match in matches:
    print(match.text)
```
